```python
import pandas as pd

# Specify the path to the xlsx file
xlsx_file_path = 'Netherland_footprint.xlsx'

# Read the xlsx file into a dataframe
Footprint_df = pd.read_excel(xlsx_file_path)

# Make index and remove the quote signs from the index
names = Footprint_df.iloc[:,1].str.strip("'")
Footprint_df.index = names

# rename columns
Footprint_df = Footprint_df.rename(columns={'GHG (kg CO2 eq)': 'Footprint_GHG', 'CO2 (Gg)': 'Footprint_CO2'})


# Display the dataframe
Footprint_df
```

```python
# Group by the product index and sum the values
sum_per_product = Footprint_df.groupby(Footprint_df.index)['Footprint_GHG'].sum()

# sum_per_product = Footprint_df.groupby(Footprint_df.index)['Footprint_GHG'].sum()
sum_per_product = sum_per_product.reindex(Footprint_df.index[0:200]) # reindex
```

```python
# Export the ordered_series as a CSV file
sum_per_product.to_csv('EXIOBASE_footprints_sum_Jiamin.csv', header=True)
sum_per_product.to_csv('EXIOBASE_footprints_sum_EXCEL_Jiamin.csv', header=True, sep = ";", decimal = ",")

print("Series exported to 'EXIOBASE_footprints_sum_Jiamin.csv' and 'EXIOBASE_footprints_sum_EXCEL_Jiamin.csv'")
```
